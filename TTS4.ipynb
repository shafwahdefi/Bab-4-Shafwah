```python
%%writefile panen.py

def hitung_total(panen):
    return sum(panen)


def hitung_diskon(total, persen_diskon):
    return total * (persen_diskon / 100)


def tampilkan_laporan(panen):
    total = hitung_total(panen)
    diskon = hitung_diskon(total, 10)
    total_setelah_diskon = total - diskon

    print("=== LAPORAN HASIL PANEN ===")
    print(f"Total hasil panen: {total} kg")
    print(f"Diskon 10%: {diskon} kg")
    print(f"Total setelah diskon: {total_setelah_diskon} kg")


if __name__ == "__main__":
    data_panen = [120, 150, 100]
    tampilkan_laporan(data_panen)
```
